## Masque directionnel:

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

img = cv2.imread('cameraman.tif')
img = img.astype(np.float32) / 255.0

hx = np.array([[0, -1, 0], [0, 1, 0], [0, 0, 0]])
hy = np.array([[0, 0, 0], [-1, 1, 0], [0, 0, 0]])
dxu = cv2.filter2D(img, -1, hx)
dyu = cv2.filter2D(img, -1, hy)
ndu = np.sqrt(dxu**2 + dyu**2)
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(dxu, cmap='gray')
plt.title('derive partielle par rapport à x')
plt.subplot(1, 3, 2)
plt.imshow(dyu, cmap='gray')
plt.title('derive partielle par rapport à y')
plt.subplot(1, 3, 3)
plt.imshow(ndu, cmap='gray')
plt.title('gradient directionnel')
plt.show()
```

## masque de Roberts 

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

img = cv2.imread('rice.png')
img = img.astype(np.float32) / 255.0

hx = np.array([[0, 0,0], [0, 0, 1], [0, -1, 0]])
hy = np.array([[0, 0, 0], [0, -1, 0], [0, 0, 1]])
dxu = cv2.filter2D(img, -1, hx)
dyu = cv2.filter2D(img, -1, hy)
ndu = np.sqrt(dxu**2 + dyu**2)
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(dxu, cmap='gray')
plt.title('derive oblique horizontal')
plt.subplot(1, 3, 2)
plt.imshow(dyu, cmap='gray')
plt.title('derive oblique vertical')
plt.subplot(1, 3, 3)
plt.imshow(ndu, cmap='gray')
plt.title('norme du gradient ')
plt.show()
```

## Masque de Prewitt

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

img = cv2.imread('rice.png')
img = img.astype(np.float32) / 255.0

hx = np.array([[-1, 0 , 1], [-1,0 ,1], [-1 ,0,1]])/3
hy = np.array([[-1 , -1,-1], [0, 0, 0], [1, 1, 1]])/3
dxu = cv2.filter2D(img, -1, hx)
dyu = cv2.filter2D(img, -1, hy)
ndu = np.sqrt(dxu**2 + dyu**2)
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(dxu, cmap='gray')
plt.subplot(1, 3, 2)
plt.imshow(dyu, cmap='gray')
plt.subplot(1, 3, 3)
plt.imshow(ndu, cmap='gray')
plt.title('norme L2 du gradient')
plt.show()
```

## Seuillage du gradient :


```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

img = cv2.imread('cameraman.tif', cv2.IMREAD_GRAYSCALE)
img = img.astype(np.float32) / 255.0

hsx  = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0 ,1]])/4
hsy = np.array([[-1,-2, -1], [0 , 0, 0], [ 1 ,2, 1]])/4

dxu = cv2.filter2D(img, -1, hsx)
dyu = cv2.filter2D(img, -1, hsy)


ndu = np.sqrt(dxu**2 + dyu**2)
ndu_uint8 = (ndu * 255).astype(np.uint8)

_, Su = cv2.threshold(ndu_uint8, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
plt.subplot(121)
plt.imshow(ndu, cmap='gray')
plt.title('Gradient')

plt.subplot(122)
plt.imshow(Su, cmap='gray')
plt.title('Seuillage Otsu')

plt.show()
```

## Seuillage par annulation de laplacien:

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

img = cv2.imread('cameraman.tif', cv2.IMREAD_GRAYSCALE)
img = img.astype(np.float32) / 255.0
m,n = img.shape[0], img.shape[1]
# print(m,n)
D = np.array([[0, 1, 0], [1, -4, 1], [0, 1 ,0]])

Du = cv2.filter2D(img, -1, D)
#Otsu travaille sur uint8
Du_uint8 = cv2.normalize(Du, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

v = Du.copy()
eps, _ = cv2.threshold(Du_uint8, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
eps = eps / 255.0 #l echelle de eps doit etre la meme que celle de Du
for i in range(m):
    for j in range(n):
        if Du[i,j] > eps:
            v[i,j] = 1
        elif Du[i,j] < -eps:
            v[i,j] = -1
        elif -eps <= Du[i,j] <= eps:
            v[i,j] = 0
v_uint8 = cv2.normalize(v, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)
seuil, Su = cv2.threshold(v_uint8, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.imshow(img, cmap='gray')
plt.title('Original')
plt.subplot(1, 3, 2)
plt.imshow(Du, cmap='gray')
plt.subplot(1, 3, 3)
plt.imshow(Su, cmap='gray')
plt.show()
```

## Detection de Canny des Conteurs :

## Visualisation par surface 3D :

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
from skimage import random_noise

img = cv2.imread('cameraman.tif')
img = img.astype(np.float32) / 255.0

IB =random_noise(img, mode='gaussian', var=0.01)

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original')
plt.subplot(1, 2, 2)
plt.imshow(IB, cmap='gray')
plt.title('Noisy Image')
plt.colormaps['gray']

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

I = cv2.imread('cameraman.tif', cv2.IMREAD_GRAYSCALE)
I = I.astype(np.float32) / 255.0  # Equivalent to im2double

IB = random_noise(I, mode='gaussian', var=0.01)

fig = plt.figure(figsize=(10, 8))
plt.subplot(2,2,1)
plt.imshow(I, cmap='gray')
plt.title('Image originale')
plt.axis('off')

plt.subplot(2,2,2)
plt.imshow(IB, cmap='gray')
plt.title('Image bruitée')
plt.axis('off')

x = np.arange(I.shape[1])
y = np.arange(I.shape[0])
X, Y = np.meshgrid(x, y)

ax1 = fig.add_subplot(2,2,3, projection='3d')
ax1.plot_surface(X, Y, I, cmap='viridis')
ax1.set_title('Surface originale')
ax1.view_init(45, 60)

ax2 = fig.add_subplot(2,2,4, projection='3d')
ax2.plot_surface(X, Y, IB, cmap='viridis')
ax2.set_title('Surface bruitée')
ax2.view_init(45, 60)

plt.tight_layout()
plt.show()
```

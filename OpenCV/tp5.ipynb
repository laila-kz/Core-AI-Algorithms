```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Déconvolution par Transformée de Fourier
def deconvolution_fourier(image_degradee, noyau, K=0.01):
    """
    Restauration d'image par déconvolution dans le domaine fréquentiel
    image_degradee : image floutée (dégradée par convolution)
    noyau : PSF (Point Spread Function) / noyau de flou
    K : constante de régularisation pour éviter la division par zéro
    """
    # Dimensions de l'image
    rows, cols = image_degradee.shape
    
    # Pad le noyau pour avoir la même taille que l'image
    pad_h = rows - noyau.shape[0]
    pad_w = cols - noyau.shape[1]
    noyau_padded = np.pad(noyau, ((0, pad_h), (0, pad_w)), 'constant')
    
    # Centrer le noyau (fftshift)
    noyau_padded = np.fft.ifftshift(noyau_padded)
    
    # Transformée de Fourier de l'image dégradée et du noyau
    F_image = np.fft.fft2(image_degradee)
    F_noyau = np.fft.fft2(noyau_padded)
    
    # Éviter la division par zéro avec régularisation
    F_noyau = F_noyau + K
    
    # Déconvolution : division dans le domaine fréquentiel
    F_resultat = F_image / F_noyau
    
    # Transformée inverse pour revenir dans le domaine spatial
    image_restauree = np.fft.ifft2(F_resultat)
    image_restauree = np.abs(image_restauree)
    
    # Normalisation
    image_restauree = cv2.normalize(image_restauree, None, 0, 255, cv2.NORM_MINMAX)
    
    return np.uint8(image_restauree)

# Programme principal
if __name__ == "__main__":
    # Charger l'image
    img = cv2.imread('cameraman.tif', 0)
    img = img.astype(np.float32) / 255.0
    
    # Créer un noyau de flou (moyenne 5x5) pour simuler la dégradation
    taille_noyau = 5
    noyau_flou = np.ones((taille_noyau, taille_noyau)) / (taille_noyau * taille_noyau)
    
    # Simuler l'image dégradée par convolution
    img_degradee = cv2.filter2D(img, -1, noyau_flou)
    
    # Ajouter un peu de bruit
    bruit = np.random.normal(0, 0.01, img.shape)
    img_degradee = img_degradee + bruit
    img_degradee = np.clip(img_degradee, 0, 1)
    
    # Appliquer la déconvolution
    img_restauree = deconvolution_fourier(img_degradee, noyau_flou, K=0.001)
    
    # Affichage des résultats
    plt.figure(figsize=(15, 5))
    
    plt.subplot(131)
    plt.imshow(img, cmap='gray')
    plt.title('Image originale')
    plt.axis('off')
    
    plt.subplot(132)
    plt.imshow(img_degradee, cmap='gray')
    plt.title('Image dégradée (flou + bruit)')
    plt.axis('off')
    
    plt.subplot(133)
    plt.imshow(img_restauree, cmap='gray')
    plt.title('Image restaurée (Déconvolution)')
    plt.axis('off')
    
    plt.tight_layout()
    plt.show()
    
    # Sauvegarder les résultats
    cv2.imwrite('image_originale.png', (img * 255).astype(np.uint8))
    cv2.imwrite('image_degradee.png', (img_degradee * 255).astype(np.uint8))
    cv2.imwrite('image_restauree.png', img_restauree)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def variance_locale(region):
    """
    Calcul de la variance d'une région (matrice)
    """
    moyenne = np.mean(region)
    variance = np.mean((region - moyenne) ** 2)
    return variance

def filtre_nagmod(image, taille_fenetre=5):
    """
    Filtre de Nagao modifié (Nagmod) pour éliminer le bruit de Poisson
    tout en préservant les contours
    
    Principe : Pour chaque pixel, on regarde 9 fenêtres carrées autour du pixel,
    on calcule la variance de chaque fenêtre, on garde celle avec la plus 
    faible variance et on remplace le pixel par la moyenne de cette fenêtre.
    """
    rows, cols = image.shape
    demi_taille = taille_fenetre // 2
    
    # Image de sortie
    image_filtree = np.zeros_like(image, dtype=np.float32)
    
    # Padding de l'image pour gérer les bords (technique du miroir)
    image_padded = cv2.copyMakeBorder(image, demi_taille*2, demi_taille*2, 
                                       demi_taille*2, demi_taille*2, 
                                       cv2.BORDER_REFLECT)
    
    # Définir les 9 fenêtres de Nagmod (5x5 avec différentes positions)
    # Chaque fenêtre est définie par (y_offset, x_offset) du coin supérieur gauche
    offsets_fenetres = [
        (0, 0),      # Haut-gauche
        (0, demi_taille),  # Haut-centre
        (0, demi_taille*2), # Haut-droite
        (demi_taille, 0),   # Milieu-gauche
        (demi_taille, demi_taille),  # Centre (fenêtre centrale)
        (demi_taille, demi_taille*2), # Milieu-droite
        (demi_taille*2, 0),  # Bas-gauche
        (demi_taille*2, demi_taille),  # Bas-centre
        (demi_taille*2, demi_taille*2) # Bas-droite
    ]
    
    # Parcourir chaque pixel de l'image originale
    for i in range(rows):
        for j in range(cols):
            # Position dans l'image paddée
            pi = i + demi_taille * 2
            pj = j + demi_taille * 2
            
            variances = []
            moyennes = []
            
            # Calculer variance et moyenne pour chacune des 9 fenêtres
            for dy, dx in offsets_fenetres:
                # Extraire la fenêtre
                fenetre = image_padded[pi + dy - demi_taille : pi + dy + demi_taille + 1,
                                        pj + dx - demi_taille : pj + dx + demi_taille + 1]
                
                var = variance_locale(fenetre)
                moy = np.mean(fenetre)
                
                variances.append(var)
                moyennes.append(moy)
            
            # Trouver l'index de la variance minimale
            idx_min_var = np.argmin(variances)
            
            # Affecter la moyenne de la région à variance minimale
            image_filtree[i, j] = moyennes[idx_min_var]
    
    # Normalisation
    image_filtree = np.clip(image_filtree, 0, 255)
    
    return image_filtree.astype(np.uint8)

def ajouter_bruit_poisson(image, scale=1.0):
    """
    Ajoute un bruit de Poisson à l'image
    Le bruit de Poisson est typique des images médicales et astronomiques
    """
    # Normaliser l'image
    vals = len(np.unique(image))
    vals = 2 ** np.ceil(np.log2(vals))
    
    # Générer le bruit de Poisson
    noisy = np.random.poisson(image * vals / scale) / float(vals) * scale
    
    return np.clip(noisy, 0, 255).astype(np.uint8)

# Programme principal
if __name__ == "__main__":
    # Charger l'image
    I = cv2.imread('cameraman.tif', 0)
    I = I.astype(np.float32)
    
    # Normaliser pour le bruit de Poisson
    I_normalisee = I / 255.0
    
    # Ajouter le bruit de Poisson
    AA = ajouter_bruit_poisson(I, scale=1.0)
    
    # Appliquer le filtre de Nagmod
    # Utilisation d'une fenêtre 5x5 comme dans le cours
    AB = filtre_nagmod(AA, taille_fenetre=5)
    
    # Affichage des résultats
    plt.figure(figsize=(15, 5))
    
    plt.subplot(131)
    plt.imshow(I, cmap='gray')
    plt.title('Image originale')
    plt.axis('off')
    
    plt.subplot(132)
    plt.imshow(AA, cmap='gray')
    plt.title('Image bruitée (Bruit de Poisson)')
    plt.axis('off')
    
    plt.subplot(133)
    plt.imshow(AB, cmap='gray')
    plt.title('Image filtrée (Nagmod)')
    plt.axis('off')
    
    plt.tight_layout()
    plt.show()
    
    # Affichage des profils de ligne (comme dans le cours)
    L = I[50, :]
    LB = AA[50, :].astype(np.float32)
    LT = AB[50, :].astype(np.float32)
    
    plt.figure(figsize=(10, 6))
    plt.plot(L, 'r', label='original')
    plt.plot(LB, 'g', label='bruitée')
    plt.plot(LT, 'b', label='restaurée')
    plt.legend()
    plt.title('Profil de ligne - Comparaison')
    plt.xlabel('Position')
    plt.ylabel('Intensité')
    plt.grid(True)
    plt.show()
    
    # Sauvegarder les images
    cv2.imwrite('image_originale.png', I.astype(np.uint8))
    cv2.imwrite('image_bruitee_poisson.png', AA)
    cv2.imwrite('image_nagmod.png', AB)
```

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

img = cv2.imread('cameraman.jpeg',cv2.IMREAD_GRAYSCALE)
img = img.astype(np.float32) / 255.0
size = 15 
kernel = np.zeros((size, size))
kernel[int((size-1)/2), :] = np.ones(size)
kernel = kernel / size
motion_blur = cv2.filter2D(img, -1, kernel)

G = np.fft.fft2(motion_blur)
H = np.fft.fft2(kernel, s=img.shape)
epsilon = 1e-3
F_est = G / (H + epsilon)

f_rec = np.real(np.fft.ifft2(F_est))

plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.imshow(img, cmap='gray')
plt.title('Image originale')

plt.subplot(1,3,2)
plt.imshow(motion_blur, cmap='gray')
plt.title('Image floue')

plt.subplot(1,3,3)
plt.imshow(f_rec, cmap='gray')
plt.title('Image reconstruite par FFT')
plt.show()
```

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

img = cv2.imread('cameraman.jpeg', cv2.IMREAD_GRAYSCALE)
img = img.astype(np.float32) / 255.0
noisy = np.random.poisson(img * 255.0) / 255.0
noisy = noisy.astype(np.float32)

def nagao_filter(image):
    h, w = image.shape
    result = np.zeros_like(image)


    offsets = [
        (-2, -2), (-2, 0), (-2, 2),
        (0, -2),  (0, 0),  (0, 2),
        (2, -2),  (2, 0),  (2, 2)
    ]

    for i in range(2, h-2):
        for j in range(2, w-2):
            min_var = 1e10
            best_mean = 0

            for dx, dy in offsets:
                sub = image[i+dx:i+dx+3, j+dy:j+dy+3]

                mean = np.mean(sub)
                var = np.var(sub)

                if var < min_var:
                    min_var = var
                    best_mean = mean

            result[i, j] = best_mean

    return result

filtered = nagao_filter(noisy)

plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.imshow(img, cmap='gray')
plt.title('Image originale')

plt.subplot(1,3,2)
plt.imshow(noisy, cmap='gray')
plt.title('Image bruitée (Poisson)')

plt.subplot(1,3,3)
plt.imshow(filtered, cmap='gray')
plt.title('Image restaurée (Nagao)')
plt.show()
```
